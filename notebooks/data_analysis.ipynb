```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
import numpy as np
```

```python
df = pd.read_csv("../data/engineered/engineered_dataset.csv")
```

```python
df.info()
```

```python
plt.figure(figsize=(10,6))
sns.countplot(y='Genre', data=df, order=df['Genre'].value_counts().index, palette='viridis')
plt.title("Distribution of Movie Genres")
plt.xlabel("Count")
plt.ylabel("Genre")
plt.tight_layout()
plt.savefig('../visuals/Genre_Distribution.png')
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.histplot(df['Duration_min'], bins=30, kde=True, color='skyblue')
plt.title("Distribution of Movie Durations (min)")
plt.xlabel("Duration_min")
plt.ylabel("Count")
plt.tight_layout()
plt.savefig('../visuals/Duration_Distribution.png')
plt.show()
```

```python
df['IMDb_Rating'] = df['IMDb_Rating'].astype(str).str.replace(',', '.').astype(float)
df['Votes'] = df['Votes'].astype(str).str.replace(r'[^\d]', '', regex=True).astype(float)
df['Duration_min'] = pd.to_numeric(df['Duration_min'], errors='coerce')
df['Description_length'] = pd.to_numeric(df['Description_length'], errors='coerce')

numeric_features = ['Duration_min', 'IMDb_Rating', 'Votes', 'Description_length']
plt.figure(figsize=(10,6))
sns.heatmap(df[numeric_features].corr(), annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Correlation Heatmap of Numeric Features")
plt.tight_layout()
plt.savefig('../visuals/Numeric_Correlation_Heatmap.png')
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.histplot(df['Description_length'], bins=30, kde=True, color='orange')
plt.title("Distribution of Description_length")
plt.xlabel("Description_length")
plt.ylabel("Count")
plt.tight_layout()
plt.savefig('../visuals/Description_length_Distribution.png')
plt.show()
```

```python
# Rated
plt.figure(figsize=(10,6))
sns.countplot(y='Rated', data=df, order=df['Rated'].value_counts().index, palette='viridis')
plt.title("Distribution of Rated")
plt.xlabel("Count")
plt.ylabel("Rated")
plt.tight_layout()
plt.savefig('../visuals/Rated_Distribution.png')
plt.show()
```

```python
# Duration_category
plt.figure(figsize=(10,6))
sns.countplot(y='Duration_category', data=df, order=df['Duration_category'].value_counts().index, palette='viridis')
plt.title("Distribution of Duration_category")
plt.xlabel("Count")
plt.ylabel("Duration_category")
plt.tight_layout()
plt.savefig('../visuals/Duration_category_Distribution.png')
plt.show()
```

```python
# IMDb Rating distribution
plt.figure(figsize=(10,6))
sns.histplot(df['IMDb_Rating'], bins=20, kde=True, color='purple')
plt.title("Distribution of IMDb Ratings")
plt.xlabel("IMDb_Rating")
plt.ylabel("Count")
plt.tight_layout()
plt.savefig('../visuals/IMDb_Rating_Distribution.png')
plt.show()
```

```python
# IMDb Rating average by Genre (barplot)
plt.figure(figsize=(12,6))
sns.barplot(x='IMDb_Rating', y='Genre', data=df, ci=None, palette='Set2')
plt.title("Average IMDb Rating by Genre")
plt.xlabel("IMDb_Rating")
plt.ylabel("Genre")
plt.tight_layout()
plt.savefig('../visuals/Average_IMDb_by_Genre.png')
plt.show()
```

```python
# Votes average by Genre (barplot)
plt.figure(figsize=(12,6))
sns.barplot(x='Votes', y='Genre', data=df, ci=None, palette='coolwarm')
plt.title("Average Votes by Genre")
plt.xlabel("Votes")
plt.ylabel("Genre")
plt.tight_layout()
plt.savefig('../visuals/Average_Votes_by_Genre.png')
plt.show()
```

```python
# Duration_category count by Genre (stacked bar alternative)
durcat_genre = df.groupby(['Genre','Duration_category']).size().unstack(fill_value=0)
durcat_genre.plot(kind='barh', stacked=True, figsize=(12,6), colormap='cool')
plt.title("Duration Category Count by Genre")
plt.xlabel("Count")
plt.ylabel("Genre")
plt.tight_layout()
plt.savefig('../visuals/DurationCategory_by_Genre.png')
plt.show()
```

```python
import plotly.express as px

fig = px.scatter(
    df,
    x='Duration_min',              
    y='IMDb_Rating',               
    size='Description_length',      
    color='Genre',                  
    hover_data=['Votes', 'Stars', 'Rated'],  
    title="Interactive Bubble: Duration vs IMDb Rating by Genre"
)

fig.update_layout(
    xaxis_title="Duration (min)",
    yaxis_title="IMDb Rating",
    legend_title="Genre"
)

fig.show()
```
