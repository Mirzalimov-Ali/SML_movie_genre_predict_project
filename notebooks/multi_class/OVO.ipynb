```python
import pandas as pd
from sklearn.model_selection import train_test_split, KFold, cross_val_score
from sklearn.metrics import accuracy_score, classification_report
import numpy as np
import os
```

```python
os.chdir(r'C:\SML_Projects\SML_project_5')
```

```python
from sklearn.multiclass import OneVsOneClassifier
```

```python
df = pd.read_csv('data/preprocessed/preprocessed_dataset.csv')
```

```python
df.info()
```

```python
df.drop('Id', axis=1, inplace=True)
```

```python
x = df.drop('Genre', axis=1)
y = df['Genre']

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
kf = KFold(n_splits=5, shuffle=True, random_state=42)
```

# LinearRegression

```python
from sklearn.linear_model import LogisticRegression

lr = OneVsOneClassifier(LogisticRegression(max_iter=1000))
lr.fit(x_train, y_train)

y_pred = lr.predict(x_test)

lr_score = accuracy_score(y_test, y_pred)
lr_scores = cross_val_score(lr, x, y, cv=kf, scoring='f1_macro')

print("Linear Regression R2:", lr_score)
print("K-Fold mean:", lr_scores.mean())
print("K-Fold std:", lr_scores.std())
```

# Decision Tree

```python
from sklearn.tree import DecisionTreeClassifier

dt = DecisionTreeClassifier(random_state=42)
dt.fit(x_train, y_train)

importance = dt.feature_importances_
selected_features = x_train.columns[importance > np.mean(importance)]

dt.fit(x_train[selected_features], y_train)
y_pred = dt.predict(x_test[selected_features])

dt_score = accuracy_score(y_test, y_pred)
dt_scores = cross_val_score(dt, x[selected_features], y, cv=kf, scoring='f1_macro')

print("Decision Tree R2:", dt_score)
print("K-Fold mean:", dt_scores.mean())
print("K-Fold std:", dt_scores.std())
```

# Random Forest

```python
from sklearn.ensemble import RandomForestClassifier

rf = RandomForestClassifier(n_estimators=200, random_state=42)
rf.fit(x_train, y_train)

importance = rf.feature_importances_
selected_features = x_train.columns[importance > np.mean(importance)]

rf.fit(x_train[selected_features], y_train)
y_pred = rf.predict(x_test[selected_features])

rf_score = accuracy_score(y_test, y_pred)
rf_scores = cross_val_score(rf, x[selected_features], y, cv=kf, scoring='f1_macro')

print("Random Forest R2:", rf_score)
print("K-Fold mean:", rf_scores.mean())
print("K-Fold std:", rf_scores.std())
```

# Gradient Boosting

```python
from sklearn.ensemble import GradientBoostingClassifier

gbr = GradientBoostingClassifier(random_state=42)
gbr.fit(x_train, y_train)
    
importance = gbr.feature_importances_
selected_features = x_train.columns[importance > np.mean(importance)]

gbr.fit(x_train[selected_features], y_train)
y_pred = gbr.predict(x_test[selected_features])

gbr_score = accuracy_score(y_test, y_pred)
gbr_scores = cross_val_score(gbr, x[selected_features], y, cv=kf, scoring='f1_macro')

print("Gradient Boosting R2:", gbr_score)
print("K-Fold mean:", gbr_scores.mean())
print("K-Fold std:", gbr_scores.std())
```

# Extra Tree

```python
from sklearn.ensemble import ExtraTreesClassifier

et = ExtraTreesClassifier(random_state=42)
et.fit(x_train, y_train)

importance = et.feature_importances_
selected_features = x_train.columns[importance > np.mean(importance)]

et.fit(x_train[selected_features], y_train)
y_pred = et.predict(x_test[selected_features])

et_score = accuracy_score(y_test, y_pred)
et_scores = cross_val_score(et, x[selected_features], y, cv=kf, scoring='f1_macro')

print("Extra Trees R2:", et_score)
print("K-Fold mean:", et_scores.mean())
print("K-Fold std:", et_scores.std())
```

# Hist Gradient Boosting

```python
from sklearn.ensemble import HistGradientBoostingClassifier

hgb = HistGradientBoostingClassifier(random_state=42)
hgb.fit(x_train, y_train)

y_pred = hgb.predict(x_test)

hgb_score = accuracy_score(y_test, y_pred)
hgb_scores = cross_val_score(hgb, x, y, cv=kf, scoring='f1_macro')

print("HistGradientBoosting R2:", hgb_score)
print("K-Fold mean:", hgb_scores.mean())
print("K-Fold std:", hgb_scores.std())
```

# KNN

```python
from sklearn.neighbors import KNeighborsClassifier

knn = OneVsOneClassifier(KNeighborsClassifier(n_neighbors=10))

knn.fit(x_train, y_train)
y_pred = knn.predict(x_test)

knn_score = accuracy_score(y_test, y_pred)
knn_scores = cross_val_score(knn, x, y, cv=kf, scoring='f1_macro')

print("KNN R2:", knn_score)
print("K-Fold mean:", knn_scores.mean())
print("K-Fold std:", knn_scores.std())
```

# Adaboost

```python
from sklearn.ensemble import AdaBoostClassifier

ada = AdaBoostClassifier(random_state=42)
ada.fit(x_train, y_train)

importance = ada.feature_importances_
selected_features = x_train.columns[importance > np.mean(importance)]

ada.fit(x_train[selected_features], y_train)
y_pred = ada.predict(x_test[selected_features])

ada_score = accuracy_score(y_test, y_pred)
ada_scores = cross_val_score(ada, x[selected_features], y, cv=kf, scoring='f1_macro')

print("AdaBoost R2:", ada_score)
print("K-Fold mean:", ada_scores.mean())
print("K-Fold std:", ada_scores.std())
```

# XGBoost

```python
from xgboost import XGBClassifier

xgb = XGBClassifier(random_state=42)
xgb.fit(x_train, y_train)

importance = xgb.feature_importances_
selected_features = x_train.columns[importance > np.mean(importance)]

xgb.fit(x_train[selected_features], y_train)
y_pred = xgb.predict(x_test[selected_features])

xgb_score = accuracy_score(y_test, y_pred)
xgb_scores = cross_val_score(xgb, x[selected_features], y, cv=kf, scoring='f1_macro')

print("XGBoost R2:", xgb_score)
print("K-Fold mean:", xgb_scores.mean())
print("K-Fold std:", xgb_scores.std())
```

# LightGBM

```python
from lightgbm import LGBMClassifier

lgbm = LGBMClassifier(random_state=42)
lgbm.fit(x_train, y_train)

importance = lgbm.feature_importances_
selected_features = x_train.columns[importance > np.mean(importance)]

lgbm.fit(x_train[selected_features], y_train)
y_pred = lgbm.predict(x_test[selected_features])

lgbm_score = accuracy_score(y_test, y_pred)
lgbm_scores = cross_val_score(lgbm, x[selected_features], y, cv=kf, scoring='f1_macro')

print("LightGBM R2:", lgbm_score)
print("K-Fold mean:", lgbm_scores.mean())
print("K-Fold std:", lgbm_scores.std())
```

# CatBoost

```python
from catboost import CatBoostClassifier

cat = CatBoostClassifier(iterations=200, depth=6, learning_rate=0.1, verbose=0, random_state=42)
cat.fit(x_train, y_train)

importance = cat.feature_importances_
selected_features = x_train.columns[importance > np.mean(importance)]

cat.fit(x_train[selected_features], y_train)
y_pred = cat.predict(x_test[selected_features])

cat_score = accuracy_score(y_test, y_pred)
cat_scores = cross_val_score(cat, x[selected_features], y, cv=kf, scoring='f1_macro')

print("LightGBM R2:", cat_score)
print("K-Fold mean:", cat_scores.mean())
print("K-Fold std:", cat_scores.std())
```

# Bagging

```python
from sklearn.ensemble import BaggingClassifier
bagging = BaggingClassifier(random_state=42)

bagging.fit(x_train, y_train)
y_pred = bagging.predict(x_test)

bag_score = accuracy_score(y_test, y_pred)
bag_scores = cross_val_score(bagging, x, y, cv=kf, scoring='f1_macro')

print(f'Bagging Classifier score: {bag_score}')
print("K-Fold mean:", bag_scores.mean())
print("K-Fold std:", bag_scores.std())
print("Selected features:", list(selected_features))
print("\nClassification Report:\n", classification_report(y_test, y_pred))
```

```python
from rich.table import Table
from rich.console import Console

console = Console()

results = [
    ['Logistic Regression', lr_score, lr_scores.mean(), lr_scores.std()],
    ['Decision Tree', dt_score, dt_scores.mean(), dt_scores.std()],
    ['Random Forest', rf_score, rf_scores.mean(), rf_scores.std()],
    ['Extra Trees', et_score, et_scores.mean(), et_scores.std()],
    ['Gradient Boosting', gbr_score, gbr_scores.mean(), gbr_scores.std()],
    ['HistGradientBoosting', hgb_score, hgb_scores.mean(), hgb_scores.std()],
    ['AdaBoost', ada_score, ada_scores.mean(), ada_scores.std()],
    ['XGBoost', xgb_score, xgb_scores.mean(), xgb_scores.std()],
    ['LightGBM', lgbm_score, lgbm_scores.mean(), lgbm_scores.std()],
    ['CatBoost', cat_score, cat_scores.mean(), cat_scores.std()],
    ['KNN', knn_score, knn_scores.mean(), knn_scores.std()],
    ['Bagging', bag_score, bag_scores.mean(), bag_scores.std()],
]

result_sorted = sorted(results, key=lambda i: i[1], reverse=True)

best_model = max(results, key=lambda x: x[1])
worst_model = min(results, key=lambda x: x[1])

table = Table(title="OVO Models Comparison", show_lines=True)
table.add_column("Algorithm")
table.add_column("Accuracy")
table.add_column("K-Fold mean", justify="right")
table.add_column("K-Fold std", justify="right")

for row in result_sorted:
    algo, accuracy, kmean, kstd = row
    if row == best_model:
        table.add_row(f"[bold green]{algo}[/bold green]",
                      f"[bold green]{accuracy:.2f}[/bold green]",
                      f"[bold green]{kmean:.2f}[/bold green]",
                      f"[bold green]{kstd:.2f}[/bold green]")
    elif row == worst_model:
        table.add_row(f"[bold salmon1]{algo}[/bold salmon1]",
                      f"[bold salmon1]{accuracy:.2f}[/bold salmon1]",
                      f"[bold salmon1]{kmean:.2f}[/bold salmon1]",
                      f"[bold salmon1]{kstd:.2f}[/bold salmon1]")
    else:
        table.add_row(algo, f"{accuracy:.2f}", f"{kmean:.2f}", f"{kstd:.2f}")

console.print(table)
```
